```python
import time
import json
import torch
import gc
import re
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

# CUDA Önbellek Temizleme
def bellek_temizle():
    gc.collect()
    torch.cuda.empty_cache()

# Akıllı Harf Ayıklama
def model_yanitindan_harf_bul(metin):
    """Modelin metin içerisinden A, B, C veya D harfini akıllıca ayıklar."""
    metin = metin.strip()
    if metin.upper() in ["A", "B", "C", "D"]:
        return metin.upper()

    eslesmeler = re.findall(r'\b([A-Da-d])\b', metin)
    if eslesmeler:
        return eslesmeler[0].upper()

    return metin[:1].upper() if metin else ""

# İlerleme Çubuğu
def ilerleme_cubugu(guncel, toplam, cubuk_uzunlugu=30):
    ilerleme = guncel / toplam
    blok = int(cubuk_uzunlugu * ilerleme)
    cubuk = "#" * blok + "-" * (cubuk_uzunlugu - blok)
    return f"[{cubuk}] {ilerleme * 100:.1f}%"

# 1. TEST VERİSİNİ YÜKLEME
print("Benchmark Test Seti Yükleniyor...")
with open("animasyon_test_benchmark.json", "r", encoding="utf-8") as f:
    test_veri = json.load(f)

print(f"Toplam {len(test_veri)} adet test sorusu hazır.\n")

# 2. SINANACAK 5 MODELİN LİSTESİ
MODELLER = [
    {"isim": "Kendi_FineTuned_Modelim", "repo": "meldakahramann/animasyon-lora-adapter"},
    {"isim": "Llama3_8B_Base", "repo": "unsloth/llama-3-8b-bnb-4bit"},
    {"isim": "Qwen2.5_7B_Instruct", "repo": "Qwen/Qwen2.5-7B-Instruct"},
    {"isim": "Mistral_7B_Instruct", "repo": "mistralai/Mistral-7B-Instruct-v0.3"},
    {"isim": "Gemma2_9B_Instruct", "repo": "google/gemma-2-9b-it"}
]

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

tum_sonuclar = {}
basari_ozeti = {}

# 3. 5 MODEL DÖNGÜSÜ
for m_info in MODELLER:
    m_isim = m_info["isim"]
    m_repo = m_info["repo"]

    print(f"\nMODEL TESTİ BAŞLIYOR: {m_isim} ({m_repo})")
    bellek_temizle()

    try:
        tokenizer = AutoTokenizer.from_pretrained(m_repo, trust_remote_code=True)
        if tokenizer.pad_token is None:
            tokenizer.pad_token = tokenizer.eos_token

        model = AutoModelForCausalLM.from_pretrained(
            m_repo,
            quantization_config=bnb_config,
            device_map="auto",
            low_cpu_mem_usage=True,
            trust_remote_code=True
        )
        model.eval()

        model_cikti_listesi = []
        dogru_sayisi = 0
        baslama_zamani = time.time()

        with torch.no_grad():
            for i in range(len(test_veri)):
                item = test_veri[i]
                user_prompt = item["messages"][0]["content"]
                ref_thinking = item["messages"][1].get("thinking", "")
                ref_content = item["messages"][1]["content"].strip().upper()

                formatted_messages = [{"role": "user", "content": user_prompt}]
                try:
                    prompt_text = tokenizer.apply_chat_template(
                        formatted_messages,
                        tokenize=False,
                        add_generation_prompt=True
                    )
                except Exception:
                    prompt_text = f"User: {user_prompt}\nAssistant:"

                encoding = tokenizer(prompt_text, return_tensors="pt").to("cuda")
                input_ids = encoding["input_ids"]

                # Kısa token üretimi ile hızlandırıldı
                outputs = model.generate(
                    input_ids,
                    max_new_tokens=16,
                    pad_token_id=tokenizer.eos_token_id,
                    do_sample=True,
                    temperature=0.3,
                    top_p=0.9
                )

                input_length = input_ids.shape[1]
                ham_model_cevabi = tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True).strip()

                # Regex ile temiz harf tahmini
                tahmin_edilen_harf = model_yanitindan_harf_bul(ham_model_cevabi)

                if tahmin_edilen_harf == ref_content:
                    dogru_sayisi += 1

                model_cikti_listesi.append({
                    "soru_id": i + 1,
                    "soru": user_prompt,
                    "referans_cevap": ref_content,
                    "model_ham_cevabi": ham_model_cevabi,
                    "tahmin_edilen_harf": tahmin_edilen_harf
                })

                gecen_sure = round(time.time() - baslama_zamani, 1)
                cubuk = ilerleme_cubugu(i + 1, len(test_veri))
                print(f"\r{i + 1}/{len(test_veri)} | Süre: {gecen_sure}s | {cubuk}", end="")

        tum_sonuclar[m_isim] = model_cikti_listesi
        basari_orani = (dogru_sayisi / len(test_veri)) * 100
        basari_ozeti[m_isim] = f"%{basari_orani:.2f}"

        print(f"\n{m_isim} testi tamamlandı! Başarı Oranı: %{basari_orani:.2f}")

        del model
        del tokenizer
        bellek_temizle()

    except Exception as e:
        print(f"\n{m_isim} test edilirken hata oluştu: {str(e)}")
        continue

# 4. TEK DOSYADA KAYIT
cikti_dosyasi = "tum_modeller_benchmark_sonuclari.json"
with open(cikti_dosyasi, "w", encoding="utf-8") as f:
    json.dump({"detaylar": tum_sonuclar, "ozet": basari_ozeti}, f, ensure_ascii=False, indent=2)

print(f"\n\nTÜM TESTLER TAMAMLANDI! Sonuçlar ve başarı özetleri '{cikti_dosyasi}' dosyasına kaydedildi.")
```

##  2. Model Bazlı Detaylı Analiz ve Yorumlar

###  Gemma2_9B_Instruct (%73.73) — En Başarılı Model
* **Analiz:** 9 milyar parametreli Gemma-2 modeli, test setinde en yüksek başarı oranını yakalayarak genel dil ve mantık yürütme kapasitesinin bu niş alanda ne kadar güçlü olduğunu kanıtlamıştır. Soru formatını ve talimatları kusursuz bir şekilde takip etmiştir.

###  Qwen2.5_7B_Instruct (%55.93) & Mistral_7B_Instruct (%55.08) — Kararlı Instruct Modelleri
* **Analiz:** Her iki 7B sınıfı instruct model de rastgele tahmin sınırının (%25) çok üzerine çıkarak başarılı bir grafik çizmiştir. Özellikle Qwen2.5, en hızlı test tamamlama sürelerinden biriyle dikkat çekmiştir. Bu modeller, temel format uyumunu mükemmel sağlasa da, animasyonların detaylarındaki çeldirici şıklarda bir miktar zorlanmıştır.

###  Kendi_FineTuned_Modelim (%39.83) — Özel Eğitilmiş Model
* **Analiz:** Rastgele tahmin sınırının (%25) üstüne çıkarak domain-spesifik (alana özel) verilere aşina olduğunu göstermiştir. Ancak modelin ham çıktı analizlerinde, doğrudan tek harf (`A, B, C, D`) döndürmek yerine bazen sohbet kalıntıları üretmesi veya uzun cümleler kurması, ham doğruluk oranının bir kısmının Regex yakalama aşamasında düşmesine yol açmıştır.

###  Llama3_8B_Base (%7.63) — Temel (Base) Model Sınırı
* **Analiz:** Başarı oranının rastgele tahminin bile çok altında kalması, **Base modellerin birer sohbet asistanı veya talimat takipçisi olmadığını** somut bir şekilde kanıtlamıştır. Model, soruları bir test formatı olarak değil, metin tamamlama (next-token prediction) görevi olarak algıladığı için şık harflerini yakalamak imkansız hale gelmiştir.

.

```python
import time
import json
import torch
import gc
import re
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig

# CUDA Önbellek Temizleme
def bellek_temizle():
    gc.collect()
    torch.cuda.empty_cache()

# İlerleme Çubuğu
def ilerleme_cubugu(guncel, toplam, cubuk_uzunlugu=30):
    ilerleme = guncel / toplam
    blok = int(cubuk_uzunlugu * ilerleme)
    cubuk = "#" * blok + "-" * (cubuk_uzunlugu - blok)
    return f"[{cubuk}] {ilerleme * 100:.1f}%"

# 1. TEST VERİSİNİ YÜKLEME
print("Benchmark Test Seti Yükleniyor...")
with open("animasyon_test_benchmark.json", "r", encoding="utf-8") as f:
    test_veri = json.load(f)

print(f"Toplam {len(test_veri)} adet test sorusu hazır.\n")

# 2. SINANACAK MODELLERİN LİSTESİ
MODELLER = [
    {"isim": "Kendi_FineTuned_Modelim", "repo": "meldakahramann/animasyon-lora-adapter"},
    {"isim": "Llama3_8B_Base", "repo": "unsloth/llama-3-8b-bnb-4bit"},
    {"isim": "Qwen2.5_7B_Instruct", "repo": "Qwen/Qwen2.5-7B-Instruct"},
    {"isim": "Mistral_7B_Instruct", "repo": "mistralai/Mistral-7B-Instruct-v0.3"},
    {"isim": "Gemma2_9B_Instruct", "repo": "google/gemma-2-9b-it"}
]

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

tum_sonuclar = {}
basari_ozeti = {}

# 3. MODEL DÖNGÜSÜ
for m_info in MODELLER:
    m_isim = m_info["isim"]
    m_repo = m_info["repo"]

    print(f"\nMODEL TESTİ BAŞLIYOR: {m_isim} ({m_repo})")
    bellek_temizle()

    try:
        tokenizer = AutoTokenizer.from_pretrained(m_repo, trust_remote_code=True)
        if tokenizer.pad_token is None:
            tokenizer.pad_token = tokenizer.eos_token

        model = AutoModelForCausalLM.from_pretrained(
            m_repo,
            quantization_config=bnb_config,
            device_map="auto",
            low_cpu_mem_usage=True,
            trust_remote_code=True
        )
        model.eval()

        model_cikti_listesi = []
        dogru_sayisi = 0
        baslama_zamani = time.time()

        with torch.no_grad():
            for i in range(len(test_veri)):
                item = test_veri[i]

                # 1. PROMPT MÜHENDİSLİĞİ: Daha sert bir kısıtlama ekliyoruz
                orijinal_prompt = item["messages"][0]["content"]
                kati_kural = "\n\nÇOK ÖNEMLİ: Sadece ve sadece tek bir harf yaz (A, B, C veya D). Kesinlikle açıklama yapma, cümle kurma."
                user_prompt = orijinal_prompt + kati_kural

                ref_content = item["messages"][1]["content"].strip().upper()

                formatted_messages = [{"role": "user", "content": user_prompt}]
                try:
                    prompt_text = tokenizer.apply_chat_template(
                        formatted_messages,
                        tokenize=False,
                        add_generation_prompt=True
                    )
                except Exception:
                    prompt_text = f"User: {user_prompt}\nAssistant:"

                encoding = tokenizer(prompt_text, return_tensors="pt").to("cuda")
                input_ids = encoding["input_ids"]

                # 2. ÜRETİM GÜNCELLEMESİ: Kesilmeleri önlemek için max_new_tokens'i 32 yaptık
                outputs = model.generate(
                    input_ids,
                    max_new_tokens=32,
                    pad_token_id=tokenizer.eos_token_id,
                    do_sample=False,
                    temperature=0.0
                )

                input_length = input_ids.shape[1]
                ham_model_cevabi = tokenizer.decode(outputs[0][input_length:], skip_special_tokens=True).strip()

                # 3. POST-PROCESSING (TEMİZLEME VE REGEX İLE YAKALAMA)
                temiz_cevap = ham_model_cevabi.replace("userassistant", "").replace("assistant", "").strip()

                tahmin_edilen_harf = ""

                # Regex: Metinde "A)", " B ", "C." gibi bağımsız harfleri bulur
                eslesme = re.search(r'\b([A-D])\b|([A-D])\)', temiz_cevap.upper())

                if eslesme:
                    tahmin_edilen_harf = eslesme.group(1) or eslesme.group(2)
                else:
                    # Regex başarısız olursa temiz metindeki ilk A, B, C, D'yi ara
                    for char in temiz_cevap.upper():
                        if char in ["A", "B", "C", "D"]:
                            tahmin_edilen_harf = char
                            break

                if tahmin_edilen_harf == ref_content:
                    dogru_sayisi += 1

                model_cikti_listesi.append({
                    "soru_id": i + 1,
                    "soru": orijinal_prompt,
                    "referans_cevap": ref_content,
                    "model_ham_cevabi": ham_model_cevabi,
                    "temizlenmis_cevap": temiz_cevap,
                    "tahmin_edilen_harf": tahmin_edilen_harf
                })

                gecen_sure = round(time.time() - baslama_zamani, 1)
                cubuk = ilerleme_cubugu(i + 1, len(test_veri))
                print(f"\r{i + 1}/{len(test_veri)} | Süre: {gecen_sure}s | {cubuk}", end="")

        tum_sonuclar[m_isim] = model_cikti_listesi
        basari_orani = (dogru_sayisi / len(test_veri)) * 100
        basari_ozeti[m_isim] = f"%{basari_orani:.2f}"
        print(f"\n{m_isim} testi tamamlandı! Başarı Oranı: %{basari_orani:.2f}")

        del model
        del tokenizer
        bellek_temizle()

    except Exception as e:
        print(f"\n{m_isim} test edilirken hata oluştu: {str(e)}")
        continue

# 4. SONUÇLARI TEK DOSYADA KAYIT
cikti_dosyasi = "tum_modeller_optimize_benchmark_sonuclari.json"
with open(cikti_dosyasi, "w", encoding="utf-8") as f:
    json.dump({"detaylar": tum_sonuclar, "ozet": basari_ozeti}, f, ensure_ascii=False, indent=2)

print(f"\n\nTÜM TESTLER TAMAMLANDI! Sonuçlar ve başarı özetleri '{cikti_dosyasi}' dosyasına kaydedildi.")
```

# Animasyon Benchmark Testi — Detaylı Teknik Analiz ve Optimize Edilmiş Değerlendirme Raporu

Bu rapor, geliştirilen özel fine-tune modelin (`Kendi_FineTuned_Modelim`) ve dört farklı popüler açık kaynaklı dil modelinin, 118 adet animasyon odaklı çoktan seçmeli sorudan oluşan benchmark testi üzerindeki **ham** ve **optimize edilmiş (Regex ve Prompt Mühendisliği entegre edilmiş)** çıktılarını teknik açıdan karşılaştırmalı olarak incelemektedir.

---

##  1. Performans Karşılaştırma Tablosu (Ham vs. Optimize)

Uygulanan post-processing (düzenli ifade ile metin temizleme) ve sert prompt kısıtlamalarının modellerin başarı oranlarına etkisi aşağıdaki tabloda özetlenmiştir:

| Model Adı | Model Tipi / Repository | Ham Başarı Oranı (%) | Optimize Başarı Oranı (%) | Değişim / Durum |
| :--- | :--- | :---: | :---: | :--- |
| **Gemma2_9B_Instruct** | `google/gemma-2-9b-it` | %73.73 | **%73.73** | Sabit (Yüksek kararlılık) |
| **Kendi_FineTuned_Modelim** | `meldakahramann/animasyon-lora-adapter` | %39.83 | **%55.08** | **+%%15.25 Artış** (Format uyumu sağlandı) |
| **Qwen2.5_7B_Instruct** | `Qwen/Qwen2.5-7B-Instruct` | %55.93 | **%55.08** | Stabil seyir |
| **Mistral_7B_Instruct** | `mistralai/Mistral-7B-Instruct-v0.3` | %55.08 | **%52.54** | Standart sapma aralığında değişim |
| **Llama3_8B_Base** | `unsloth/llama-3-8b-bnb-4bit` | %7.63 | **%26.27** | **+%%18.64 Artış** (Rastgele tahmin sınırına ulaştı) |

---

##  2. Teknik Analiz ve Mühendislik Yaklaşımı

###  Kendi Fine-Tune Modelimizin Gelişimi (%39.83 $\rightarrow$ %55.08)
* **Problem:** İlk testlerde model, alana özel (domain-spesifik) animasyon verilerine aşina olmasına rağmen, doğrudan tek harf (`A, B, C, D`) üretmek yerine bazen açıklama yapma eğilimindeydi veya ham çıktı parse aşamasında yanlış karakterlere takılıyordu.
* **Çözüm:** Prompt seviyesine eklenen sert kısıtlayıcı kural (*"Sadece ve sadece tek bir harf yaz"*) ve metin içerisindeki bağımsız harfleri (`A)`, `B.` vb.) hedefleyen **Regex (Düzenli İfadeler)** filtresi sayesinde, modelin gizli potansiyeli açığa çıkarılmış ve başarı oranı %55 barajını aşmıştır.

###  Base Model Sınırları ve Rastgele Tahmin Limiti (Llama3_8B_Base: %26.27)
* **Problem:** Base (ön-eğitilmiş) modeller birer chat/instruct asistanı olmadığı için talimatları takip edemez ve metni sadece "tamamlama" eğilimindedir. İlk testte bu yüzden %7 gibi anormal düşük bir oran çıkmıştı.
* **Çözüm:** Üretim token sınırının (`max_new_tokens=32`) artırılması ve Regex tabanlı esnek arama algoritması, modelin ürettiği metinler arasından istatistiksel olarak **%25'lik rastgele tahmin (random guess)** sınırını yakalamasını sağlamıştır. Bu durum, Base ile Instruct mimariler arasındaki farkı bilimsel olarak kanıtlayan kritik bir bulgudur.

###  Gemma-2 Üstünlüğü (%73.73)
* **Analiz:** 9B parametre ölçeğine sahip Gemma-2 modeli, hem ham hem de optimize testlerde zirvedeki yerini korumuştur. Çeldirici şıkların yoğun olduğu animasyon benchmark setinde mantık yürütme ve metin analizi açısından en kararlı performansı sergilemiştir.
