```python
import gc
import re
import time
import torch
import pandas as pd

from peft import PeftConfig, PeftModel
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig
)

assert torch.cuda.is_available(), "Önce Colab'dan T4 GPU seçmelisin."

print("GPU:", torch.cuda.get_device_name(0))

LORA_MODEL = "haticenurcakr/turkish-classic-books-qwen-lora"

# LoRA'nın eğitildiği base modeli otomatik öğren
peft_config = PeftConfig.from_pretrained(LORA_MODEL)
BASE_MODEL = peft_config.base_model_name_or_path

MODELLER = [
    {
        "etiket": "Qwen Base",
        "model": BASE_MODEL,
        "lora": None
    },
    {
        "etiket": "Qwen + Türk Klasik Kitapları LoRA",
        "model": BASE_MODEL,
        "lora": LORA_MODEL
    },
    {
        "etiket": "SmolLM2 1.7B Instruct",
        "model": "HuggingFaceTB/SmolLM2-1.7B-Instruct",
        "lora": None
    },
    {
        "etiket": "TinyLlama 1.1B Chat",
        "model": "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
        "lora": None
    },
    {
        "etiket": "StableLM Zephyr 1.6B",
        "model": "stabilityai/stablelm-2-zephyr-1_6b",
        "lora": None
    }
]

print("Base model:", BASE_MODEL)

for ayar in MODELLER:
    print("-", ayar["etiket"], "=>", ayar["model"])
```

```python
quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

HARFLER = ["A", "B", "C", "D"]


def modeli_yukle(model_id, lora_id=None):

    tokenizer = AutoTokenizer.from_pretrained(
        model_id,
        trust_remote_code=True
    )

    tokenizer.padding_side = "left"

    if tokenizer.pad_token_id is None:
        tokenizer.pad_token = tokenizer.eos_token

    model = AutoModelForCausalLM.from_pretrained(
        model_id,
        quantization_config=quantization_config,
        device_map="auto",
        trust_remote_code=True
    )

    if lora_id is not None:
        model = PeftModel.from_pretrained(
            model,
            lora_id
        )

    model.eval()

    return model, tokenizer


def prompt_hazirla(tokenizer, soru, secenekler):

    secenek_metni = "\n".join(
        f"{HARFLER[i]}) {secenek}"
        for i, secenek in enumerate(secenekler)
    )

    istek = (
        "Aşağıdaki Türk edebiyatı sorusunu cevapla. "
        "Yalnızca doğru seçeneğin harfini yaz. "
        "Açıklama yapma.\n\n"
        f"Soru: {soru}\n"
        f"{secenek_metni}\n"
        "Cevap:"
    )

    messages = [
        {
            "role": "user",
            "content": istek
        }
    ]

    try:
        return tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=True
        )
    except Exception:
        return istek


def cevap_harfini_bul(metin):

    metin = str(metin).upper().strip()

    eslesme = re.search(
        r"(?:^|\s|[(:])([ABCD])(?:\s|$|[).,:])",
        metin
    )

    if eslesme:
        return eslesme.group(1)

    if len(metin) > 0 and metin[0] in HARFLER:
        return metin[0]

    return None


def benchmark_calistir(ayar, veri, batch_size=8):

    print("\n" + "=" * 60)
    print("Test edilen model:", ayar["etiket"])
    print("=" * 60)

    model, tokenizer = modeli_yukle(
        ayar["model"],
        ayar["lora"]
    )

    baslangic = time.time()
    kayitlar = []

    for baslangic_index in range(
        0,
        len(veri),
        batch_size
    ):

        parca = veri.iloc[
            baslangic_index:
            baslangic_index + batch_size
        ]

        promptlar = [
            prompt_hazirla(
                tokenizer,
                satir.soru,
                list(satir.secenekler)
            )
            for satir in parca.itertuples()
        ]

        inputs = tokenizer(
            promptlar,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=1024
        ).to(model.device)

        with torch.inference_mode():

            outputs = model.generate(
                **inputs,
                max_new_tokens=4,
                do_sample=False,
                pad_token_id=tokenizer.pad_token_id
            )

        yeni_tokenler = outputs[
            :,
            inputs["input_ids"].shape[1]:
        ]

        cevap_metinleri = tokenizer.batch_decode(
            yeni_tokenler,
            skip_special_tokens=True
        )

        for (_, satir), ham_cevap in zip(
            parca.iterrows(),
            cevap_metinleri
        ):

            tahmin = cevap_harfini_bul(
                ham_cevap
            )

            dogru_mu = (
                tahmin == satir["dogru_harf"]
            )

            kayitlar.append({
                "model": ayar["etiket"],
                "model_id": (
                    ayar["lora"]
                    if ayar["lora"] is not None
                    else ayar["model"]
                ),
                "soru_id": satir["id"],
                "kategori": satir["kategori"],
                "dogru_harf": satir["dogru_harf"],
                "model_tahmini": tahmin,
                "ham_cevap": ham_cevap.strip(),
                "dogru_mu": bool(dogru_mu)
            })

        tamamlanan = min(
            baslangic_index + batch_size,
            len(veri)
        )

        dogru_sayisi = sum(
            kayit["dogru_mu"]
            for kayit in kayitlar
        )

        print(
            f"\r{tamamlanan}/{len(veri)} | "
            f"Doğru: {dogru_sayisi} | "
            f"Başarı: "
            f"%{100 * dogru_sayisi / tamamlanan:.2f}",
            end=""
        )

    toplam_sure = time.time() - baslangic

    detay_df = pd.DataFrame(kayitlar)

    ozet = {
        "model": ayar["etiket"],
        "model_id": (
            ayar["lora"]
            if ayar["lora"] is not None
            else ayar["model"]
        ),
        "toplam_soru": len(detay_df),
        "dogru_cevap": int(
            detay_df["dogru_mu"].sum()
        ),
        "yanlis_cevap": int(
            (~detay_df["dogru_mu"]).sum()
        ),
        "gecersiz_cevap": int(
            detay_df["model_tahmini"].isna().sum()
        ),
        "basari_yuzdesi": round(
            100 * detay_df["dogru_mu"].mean(),
            2
        ),
        "sure_saniye": round(
            toplam_sure,
            2
        )
    }

    print(
        f"\nTamamlandı: "
        f"%{ozet['basari_yuzdesi']}"
    )

    del model
    del tokenizer

    gc.collect()
    torch.cuda.empty_cache()

    return ozet, detay_df
```

```python
def harf_tokenlerini_bul(tokenizer):

    tokenler = {}

    for harf in HARFLER:
        aday_idler = []

        # Farklı tokenizerların boşluk kullanımını destekle
        for metin in [harf, " " + harf]:

            ids = tokenizer.encode(
                metin,
                add_special_tokens=False
            )

            # Tek token olan seçenekleri kabul et
            if len(ids) == 1:
                aday_idler.append(ids[0])

        aday_idler = list(set(aday_idler))

        if len(aday_idler) == 0:
            raise ValueError(
                f"{harf} için tek token bulunamadı."
            )

        tokenler[harf] = aday_idler

    return tokenler


def benchmark_olasilikla_calistir(
    ayar,
    veri,
    batch_size=8
):

    print("\n" + "=" * 60)
    print("Test edilen model:", ayar["etiket"])
    print("=" * 60)

    model, tokenizer = modeli_yukle(
        ayar["model"],
        ayar["lora"]
    )

    harf_tokenleri = harf_tokenlerini_bul(
        tokenizer
    )

    print("Harf tokenleri:", harf_tokenleri)

    baslama_zamani = time.time()
    kayitlar = []

    for baslangic_index in range(
        0,
        len(veri),
        batch_size
    ):

        parca = veri.iloc[
            baslangic_index:
            baslangic_index + batch_size
        ]

        promptlar = [
            prompt_hazirla(
                tokenizer,
                satir.soru,
                list(satir.secenekler)
            )
            for satir in parca.itertuples()
        ]

        inputs = tokenizer(
            promptlar,
            return_tensors="pt",
            padding=True,
            truncation=True,
            max_length=1024
        ).to(model.device)

        with torch.inference_mode():

            outputs = model(
                **inputs
            )

            # Modelin sıradaki token için ürettiği puanlar
            son_token_logitleri = (
                outputs.logits[:, -1, :].float()
            )

        for satir_sirasi, (_, satir) in enumerate(
            parca.iterrows()
        ):

            harf_puanlari = {}

            for harf in HARFLER:

                token_idleri = harf_tokenleri[harf]

                puanlar = [
                    son_token_logitleri[
                        satir_sirasi,
                        token_id
                    ].item()
                    for token_id in token_idleri
                ]

                harf_puanlari[harf] = max(puanlar)

            # En yüksek olasılığa sahip harf
            tahmin = max(
                harf_puanlari,
                key=harf_puanlari.get
            )

            dogru_mu = (
                tahmin == satir["dogru_harf"]
            )

            kayitlar.append({
                "model": ayar["etiket"],
                "model_id": (
                    ayar["lora"]
                    if ayar["lora"] is not None
                    else ayar["model"]
                ),
                "soru_id": satir["id"],
                "kategori": satir["kategori"],
                "dogru_harf": satir["dogru_harf"],
                "model_tahmini": tahmin,
                "A_puani": harf_puanlari["A"],
                "B_puani": harf_puanlari["B"],
                "C_puani": harf_puanlari["C"],
                "D_puani": harf_puanlari["D"],
                "dogru_mu": bool(dogru_mu)
            })

        tamamlanan = min(
            baslangic_index + batch_size,
            len(veri)
        )

        dogru_sayisi = sum(
            kayit["dogru_mu"]
            for kayit in kayitlar
        )

        print(
            f"\r{tamamlanan}/{len(veri)} | "
            f"Doğru: {dogru_sayisi} | "
            f"Başarı: "
            f"%{100 * dogru_sayisi / tamamlanan:.2f}",
            end=""
        )

    toplam_sure = (
        time.time() - baslama_zamani
    )

    detay = pd.DataFrame(kayitlar)

    ozet = {
        "model": ayar["etiket"],
        "model_id": (
            ayar["lora"]
            if ayar["lora"] is not None
            else ayar["model"]
        ),
        "toplam_soru": len(detay),
        "dogru_cevap": int(
            detay["dogru_mu"].sum()
        ),
        "yanlis_cevap": int(
            (~detay["dogru_mu"]).sum()
        ),
        "gecersiz_cevap": 0,
        "basari_yuzdesi": round(
            100 * detay["dogru_mu"].mean(),
            2
        ),
        "sure_saniye": round(
            toplam_sure,
            2
        )
    }

    print(
        f"\nTamamlandı: "
        f"%{ozet['basari_yuzdesi']}"
    )

    del model
    del tokenizer
    del outputs

    gc.collect()
    torch.cuda.empty_cache()

    return ozet, detay
```
